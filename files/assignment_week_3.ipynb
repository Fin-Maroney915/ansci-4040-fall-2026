# Pushing your results 

```python
# Question 3.1 and 3.2: Compare 2024 vs 1965 for World
world_data = cowstats[cowstats['Entity'] == 'World']

# Get values for 1965 and 2024
cows_1965 = world_data[world_data['Year'] == 1965]['Cattle - Stocks (animals)'].values[0]
cows_2024 = world_data[world_data['Year'] == 2024]['Cattle - Stocks (animals)'].values[0]

# Calculate the difference and percentage increase
difference = cows_2024 - cows_1965
percentage_increase = ((cows_2024 - cows_1965) / cows_1965) * 100

print(f"3.1: Cows in 1965: {cows_1965:,}")
print(f"     Cows in 2024: {cows_2024:,}")
print(f"     Difference: {difference:,} more cows in 2024")
print(f"\n3.2: Percentage increase: {percentage_increase:.1f}%")
```

```python
# Question 3.3: Trend in North America
north_america_data = cowstats[cowstats['Entity'] == 'North America']

if len(north_america_data) > 0:
    # Get earliest and latest years
    earliest_year = north_america_data['Year'].min()
    latest_year = north_america_data['Year'].max()
    
    earliest_count = north_america_data[north_america_data['Year'] == earliest_year]['Cattle - Stocks (animals)'].values[0]
    latest_count = north_america_data[north_america_data['Year'] == latest_year]['Cattle - Stocks (animals)'].values[0]
    
    change = latest_count - earliest_count
    percent_change = ((latest_count - earliest_count) / earliest_count) * 100
    
    print(f"3.3: North America cattle trend ({earliest_year}-{latest_year}):")
    print(f"     {earliest_year}: {earliest_count:,} cattle")
    print(f"     {latest_year}: {latest_count:,} cattle")
    print(f"     Change: {change:,} ({percent_change:+.1f}%)")
    
    if change > 0:
        print(f"     Trend: INCREASING")
    elif change < 0:
        print(f"     Trend: DECREASING")
    else:
        print(f"     Trend: STABLE")
else:
    print("North America data not found in dataset")
```

```python
# Question 3.4: Africa cattle growth since 2000
africa_data = cowstats[cowstats['Entity'] == 'Africa']

# Get 2000 and most recent year data
africa_2000 = africa_data[africa_data['Year'] == 2000]['Cattle - Stocks (animals)'].values
africa_latest = africa_data[africa_data['Year'] == africa_data['Year'].max()]['Cattle - Stocks (animals)'].values

if len(africa_2000) > 0 and len(africa_latest) > 0:
    africa_2000_count = africa_2000[0]
    africa_latest_count = africa_latest[0]
    latest_year = africa_data['Year'].max()
    
    multiplier = africa_latest_count / africa_2000_count
    percentage_increase = ((africa_latest_count - africa_2000_count) / africa_2000_count) * 100
    
    print(f"3.4: Africa cattle growth since 2000:")
    print(f"     Year 2000: {africa_2000_count:,} cattle")
    print(f"     Year {latest_year}: {africa_latest_count:,} cattle")
    print(f"     Multiplier: {multiplier:.2f}x more cattle in {latest_year} vs 2000")
    print(f"     Percentage increase: {percentage_increase:.1f}%")
else:
    print("Africa data not found for year 2000")
```
